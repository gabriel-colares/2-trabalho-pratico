## Objetivo do Trabalho

O trabalho consolida os microdados anuais do Censo Escolar em um único CSV padronizado, valida esse arquivo com Pandera e executa consultas agregadas sobre o volume consolidado.

O fluxo principal é:

1. Ler os arquivos originais do Censo Escolar de 1995 a 2025.
2. Harmonizar nomes, tipos e categorias em 42 colunas finais.
3. Gerar `trabalho/censo-escolar.csv`.
4. Validar o CSV consolidado com um schema Pandera.
5. Executar operações agregadas em batches para grandes volumes.

## Estrutura de Pastas: Obrigatório e Opcional

Nomes obrigatórios para execução correta:

- `trabalho`: pasta onde ficam os notebooks e onde o consolidado deve ser gerado ou copiado.
- `datasets`: pasta irmã de `trabalho`, usada pelo notebook 1 para localizar os microdados originais.
- `censo-escolar.csv`: nome obrigatório do arquivo consolidado dentro de `trabalho`.
- Arquivos principais anuais: os nomes precisam seguir o padrão esperado pelo código, por exemplo `CENSOESC_1995.CSV`, `microdados_ed_basica_2007.csv` e `Tabela_Escola_2025.csv`.

Nomes opcionais ou flexíveis:

- `<pasta-raiz-do-projeto>`: pode ter qualquer nome. O importante é conter `datasets/` e `trabalho/` lado a lado.
- Subpastas dentro de `datasets`: podem ter qualquer nome, porque o notebook procura arquivos `.csv` recursivamente.
- `schema_reports` e `resultados`: são criadas automaticamente pelos notebooks quando necessário.
- `operacoes-grandes-volumes-censo-escolar`, `csv` e `graficos`: são nomes definidos pelo notebook 3 para organizar as saídas; não precisam existir antes da execução.

## Ordem de Execução dos Arquivos

Execute os notebooks nesta ordem:

1. `1-gerar-csv-censo-escolar.ipynb`
   Gera o arquivo consolidado `trabalho/censo-escolar.csv` a partir dos microdados em `datasets/`.

2. `2-schema-pandera-censo-escolar.ipynb`
   Valida o CSV consolidado com Pandera e gera os relatórios em `trabalho/schema_reports/`.

3. `3-operacoes-grandes-volumes-censo-escolar.ipynb`
   Executa consultas agregadas em batches e salva resultados em `trabalho/resultados/operacoes-grandes-volumes-censo-escolar/`.

4. `4-comparacao-desempenho-bibliotecas copy.ipynb`
   Opcional. Compara desempenho de bibliotecas usando o CSV consolidado.

## Observações para Execução Correta

- Execute os notebooks a partir da pasta `trabalho` ou da pasta raiz do projeto; os notebooks tentam detectar a raiz automaticamente.
- A pasta `datasets` deve estar no mesmo nível da pasta `trabalho`, não dentro dela.

- Os notebooks 2 e 3 dependem de `trabalho/censo-escolar.csv`.
- O processamento completo lê milhões de linhas e pode demorar alguns minutos.
